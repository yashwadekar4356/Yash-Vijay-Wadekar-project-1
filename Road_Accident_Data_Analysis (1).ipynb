# Exploratory Data Analysis 

```python


df['Year'].value_counts().sort_index().plot(kind='bar') 

plt.title('Road Accidents by Year') 
plt.xlabel('Year') 
plt.ylabel('Number of Accidents') 
plt.show()
```

```python
df['Day of Week'].value_counts().sort_index().plot(kind='bar') 

plt.title('Number of Accidents by Week') 
plt.xlabel('Day of Week') 
plt.ylabel('Number of Accidents by Week') 
plt.show()
```

```python
df['Road Condition'].value_counts().sort_index().plot(kind='bar') 

plt.title('Accidents by Road Conditions') 
plt.xlabel('Road Condition') 
plt.ylabel('Accidents') 
plt.show()
```

```python
df['Road Type'].value_counts().sort_index().plot(kind='bar') 

plt.title('Road Type Accidents') 
plt.xlabel('Road Type') 
plt.ylabel('Accidents') 
plt.show()
```

```python
sns.pairplot(df) 

plt.show()
```

```python
num = df.select_dtypes(include='number') 

sns.heatmap(num.corr(),annot=True,cmap='coolwarm') 

plt.title('Correlation Heatmap') 
plt.show()
```
